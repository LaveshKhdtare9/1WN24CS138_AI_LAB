```python
print("Lavesh Khadtare")
print("1WN24CS138")

def get_neighbors(state):
    neighbors = []

    index = state.index(0)
    row = index // 3
    col = index % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_index = new_row * 3 + new_col

            new_state = list(state)
            new_state[index], new_state[new_index] = \
                new_state[new_index], new_state[index]

            neighbors.append(tuple(new_state))

    return neighbors


def dls(state, goal, depth, path):

    if state == goal:
        return path

    if depth == 0:
        return None

    for next_state in get_neighbors(state):

        if next_state not in path:
            result = dls(
                next_state, goal, depth - 1,
                path + [next_state]
            )

            if result:
                return result

    return None


def ids(start, goal):

    depth = 0

    while True:
        result = dls(start, goal, depth, [start])

        if result:
            return result

        depth += 1


def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

solution = ids(start, goal)

print("Solution Found!")
print("Number of moves:", len(solution) - 1)

for step in solution:
    print_puzzle(step)
```
