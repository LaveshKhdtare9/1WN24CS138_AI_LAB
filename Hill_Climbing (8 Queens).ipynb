```python

print("Lavesh Khadtare")
print("1WN24CS138")

def cost(state):
    conflicts = 0

    for i in range(4):
        for j in range(i + 1, 4):
            if state[i] == state[j]:
                conflicts += 1
            elif abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def display_board(state):
    print("+---+---+---+---+")

    for row in range(4):
        print("|", end=" ")

        for col in range(4):
            if state[col] == row:
                print("Q", end=" | ")
            else:
                print(" ", end=" | ")

        print()
        print("+---+---+---+---+")


def hill_climbing():
    current = [3, 1, 2, 0]
    step = 1

    while True:
        print("\nSTEP", step)
        print("State:", current)
        print("Cost:", cost(current))
        display_board(current)

        current_cost = cost(current)
        best_state = current[:]
        best_cost = current_cost

        for i in range(4):
            for j in range(i + 1, 4):
                neighbor = current[:]
                neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

                neighbor_cost = cost(neighbor)

                print(
                    "Swap x" + str(i) + " and x" + str(j),
                    "->", neighbor,
                    "Cost:", neighbor_cost
                )

                if neighbor_cost < best_cost:
                    best_state = neighbor
                    best_cost = neighbor_cost

        if best_cost >= current_cost:
            print("\nAlgorithm terminated.")
            print("Final board:")
            display_board(current)
            print("Final cost:", current_cost)

            if current_cost == 0:
                print("Solution found!")
            else:
                print("Local minimum reached.")
            break

        current = best_state
        step += 1


hill_climbing()
```
