```python

def unify(x, y, substitution=None):
    if substitution is None:
        substitution = {}


    if x == y:
        return substitution

    if isinstance(x, str) and x.islower():
        return unify_variable(x, y, substitution)


    if isinstance(y, str) and y.islower():
        return unify_variable(y, x, substitution)

    if isinstance(x, list) and isinstance(y, list):

        if len(x) != len(y) or x[0] != y[0]:
            return None

        for a, b in zip(x[1:], y[1:]):
            substitution = unify(a, b, substitution)

            if substitution is None:
                return None

        return substitution

    return None


def unify_variable(var, value, substitution):


    if var in substitution:
        return unify(substitution[var], value, substitution)

    if isinstance(value, str) and value in substitution:
        return unify(var, substitution[value], substitution)


    if occurs(var, value, substitution):
        return None

    substitution[var] = value
    return substitution


def occurs(var, value, substitution):
    if var == value:
        return True

    if isinstance(value, list):
        return any(occurs(var, item, substitution)
                   for item in value)

    if isinstance(value, str) and value in substitution:
        return occurs(var, substitution[value], substitution)

    return False


expr1 = ["Eats", "x", "Apple"]
expr2 = ["Eats", "Lavesh", "y"]

result = unify(expr1, expr2)

print("Lavesh Khadtare")
print("1WN24CS138")

if result is not None:
    print("Unification Successful")
    print("Substitution:", result)
else:
    print("Unification Failed")
```
