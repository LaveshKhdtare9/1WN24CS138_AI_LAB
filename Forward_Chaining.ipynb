```python
#  facts
facts = {
    "barks",
    "four_legs",
    "sharp_teeth"
}

# Forward Chaining
def forward_chaining(facts):

    # Rule 1: If X barks and has four legs, X is a dog
    if "barks" in facts and "four_legs" in facts:
        facts.add("dog")
        print("Tommy is a dog")

    # Rule 2: If X has feathers and lays eggs, X is a bird
    if "feathers" in facts and "lays_eggs" in facts:
        facts.add("bird")
        print("Tommy is a bird")

    # Rule 3: If X is a dog, X is a mammal
    if "dog" in facts:
        facts.add("mammal")
        print("Tommy is a mammal")

    # Rule 4: If X is a mammal and has sharp teeth, X is a carnivore
    if "mammal" in facts and "sharp_teeth" in facts:
        facts.add("carnivore")
        print("Tommy is a carnivore")

    # Rule 5: If X is a bird, X can fly
    if "bird" in facts:
        facts.add("can_fly")
        print("Tommy can fly")

    return facts


result = forward_chaining(facts)

if "carnivore" in result:
    print("\nGoal achieved: Tommy is a Carnivore")
else:
    print("\nGoal not achieved")
```
