```python
print("Lavesh Khadtare")
print("USN:1WN24CS138")

from heapq import heappush, heappop

goal = (1, 2, 3,
        4, 0, 6,
        7, 5, 8)


def manhattan_distance(state):
    distance = 0

    for i in range(9):

        if state[i] != 0:

            # Current position
            current_row = i // 3
            current_col = i % 3

            # Goal position
            goal_index = goal.index(state[i])
            goal_row = goal_index // 3
            goal_col = goal_index % 3

            distance += abs(current_row - goal_row)
            distance += abs(current_col - goal_col)

    return distance


def get_neighbors(state):
    neighbors = []

    zero = state.index(0)

    row = zero // 3
    col = zero % 3

    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_zero = new_row * 3 + new_col

            new_state = list(state)

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            neighbors.append(tuple(new_state))

    return neighbors


def print_state(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


def a_star(start):

    # (f, g, state, path)
    pq = []

    g = 0
    h = manhattan_distance(start)
    f = g + h

    heappush(pq, (f, g, start, [start]))

    visited = set()

    while pq:

        f, g, state, path = heappop(pq)

        if state in visited:
            continue

        visited.add(state)

        if state == goal:
            return path

        for next_state in get_neighbors(state):

            if next_state not in visited:

                new_g = g + 1
                new_h = manhattan_distance(next_state)
                new_f = new_g + new_h

                heappush(
                    pq,
                    (new_f, new_g, next_state, path + [next_state])
                )

    return None

start_state = (1, 2, 3,
               4, 0, 5,
               7, 8, 6)

print("Starting State:")
print_state(start_state)

solution = a_star(start_state)

if solution:
    print(f"Solution found in {len(solution) - 1} moves:\n")
    for step_num, state in enumerate(solution):
        print(f"Step {step_num}:")
        print_state(state)
else:
    print("No solution found.")
```
