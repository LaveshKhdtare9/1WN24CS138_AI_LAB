```python
# Vacuum Cleaner Agent

print("Lavesh Khadtare")
print("USN:1WN24CS138")

room = {
    "A": "Dirty",
    "B": "Dirty"
}

position = "A"

while True:
    print("\nCurrent Room:", position)
    print("Room A:", room["A"])
    print("Room B:", room["B"])

    if room[position] == "Dirty":
        print("Action: Cleaning", position)
        room[position] = "Clean"

    elif position == "A":
        print("Action: Move from A to B")
        position = "B"

    elif position == "B":
        print("Action: Move from B to A")
        position = "A"

    if room["A"] == "Clean" and room["B"] == "Clean":
        print("\nBoth rooms are clean!")
        print("Goal completed.")
        break
```
