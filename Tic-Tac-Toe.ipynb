```python
print("Lavesh Khadtare")
print("USN:1WN24CS138")

import random

board = ["1","2","3","4","5","6","7","8","9"]

print("TIC TAC TOE")
print("You = X")
print("Computer = O")

while True:

    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])

    # Player
    move = input("Enter 1-9: ")

    if move in board:
        board[int(move)-1] = "X"
    else:
        print("Choose an empty number!")
        continue

    # Computer
    empty = []

    for x in range(9):
        if board[x] not in ["X", "O"]:
            empty.append(x)

    if len(empty) == 0:
        print("DRAW!")
        break

    computer = random.choice(empty)
    board[computer] = "O"

    print("Computer played!")

```
